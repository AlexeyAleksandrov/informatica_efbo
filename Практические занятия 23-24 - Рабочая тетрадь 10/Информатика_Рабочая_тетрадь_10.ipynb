## 1. Теоретический материал

### 1.1. Размер входа и число операций

Одну задачу можно решить разными алгоритмами. Для сравнения необходимо указать входные данные, требуемый результат и операцию, число выполнений которой будет подсчитываться.

Для поиска в списке **размер входа** $n$ — количество элементов. **Ключ поиска** — значение, которое требуется найти. Результат поиска — индекс найденного элемента или сообщение об отсутствии ключа. Индексы Python начинаются с 0, поэтому последний элемент непустого списка имеет индекс $n-1$.

В приведённых программах подсчитываются **сравнения значений**: проверка равенства элемента ключу и проверка порядка двух значений. Сравнения индексов, присваивания, обмены и создание копии списка в этот счётчик не входят. Например, проверки `values[middle] == target` и `values[middle] < target` — два отдельных сравнения. Если первая проверка успешна и функция завершилась, вторая не выполняется.

Счётчик показывает затраты выбранного типа. По нему нельзя определить точное время выполнения программы: другие операции тоже требуют времени. Однако число сравнений удобно для проверки алгоритма и изучения роста затрат при увеличении $n$.

### 1.2. Лучший и худший случаи

Число операций зависит не только от размера входа, но и от самих данных. При последовательном поиске ключ может находиться в начале списка, в конце или отсутствовать.

**Лучший случай** — допустимый вход данного размера, на котором требуется наименьшее число операций. **Худший случай** — вход того же размера с наибольшим числом операций.

Если список непустой, последовательный поиск находит первый элемент за одно сравнение. Для ключа, которого нет в списке, проверяются все $n$ элементов. Оба результата относятся к одному алгоритму, поэтому при указании сложности нужно пояснять, какой случай рассматривается.

### 1.3. Оценки O(1), O(n), O(log n), O(n²)

Обозначение $O(g(n))$ описывает верхнюю оценку порядка роста затрат при увеличении размера входа. Оно не задаёт точного числа операций и не выражается в секундах. При такой оценке постоянные множители и слагаемые меньшего порядка не определяют итоговый класс роста.

Например:

$$
C_1(n)=3n+5=O(n),
$$

$$
C_2(n)=\frac{n(n-1)}{2}=\frac{n^2-n}{2}=O(n^2).
$$

В первом выражении рост определяется $n$, во втором — $n^2$. Запись $O(n)$ не означает, что программа делает ровно $n$ действий.

| Оценка | Как растут затраты | Пример |
|---|---|---|
| $O(1)$ | Ограничены постоянной при росте $n$ | Чтение элемента списка по заданному допустимому индексу |
| $O(\log n)$ | При каждом шаге область поиска сокращается примерно вдвое | Бинарный поиск в отсортированном списке, худший случай |
| $O(n)$ | Пропорциональны размеру входа | Линейный поиск, худший случай |
| $O(n^2)$ | Пропорциональны квадрату размера входа | Сортировка выбором; сортировка вставками в худшем случае |

Обычно указывают наиболее точную из обсуждаемых оценок: она лучше характеризует алгоритм. Оценка $O$ сама по себе не означает именно худший случай; выбранный случай задаётся отдельно.

Для бинарного поиска используют логарифм по основанию 2: он связан с числом делений области пополам. Например, цепочка размеров $16,8,4,2,1$ содержит четыре уменьшения вдвое. Оставшийся единственный элемент тоже может потребовать проверки. Поэтому число итераций поиска не следует безоговорочно приравнивать к $\log_2n$.

### 1.4. Линейный поиск

**Линейный поиск** последовательно проверяет элементы, пока не найдёт ключ или не достигнет конца списка. Сортировка входа ему не требуется.

Ниже функция возвращает пару `(index, comparisons)`: индекс найденного элемента и число сравнений значений. Если ключ отсутствует, индекс равен `-1`. Этот результат означает отсутствие элемента; его не следует использовать как индекс последнего элемента списка.

```python
def linear_search(values, target):
    comparisons = 0
    for index, value in enumerate(values):
        comparisons += 1
        if value == target:
            return index, comparisons
    return -1, comparisons
```

Первое совпадение завершает поиск, поэтому при повторяющихся значениях эта функция возвращает индекс **первого** вхождения. Для пустого списка результат равен `(-1, 0)`.

В лучшем случае требуется одно сравнение, в худшем — $n$:

$$
C_{\text{лучший}}(n)=1=O(1),\qquad
C_{\text{худший}}(n)=n=O(n),\qquad n\ge1.
$$

### 1.5. Бинарный поиск

**Бинарный поиск** сравнивает ключ со средним элементом текущего участка. Он применяется к списку, упорядоченному по неубыванию: каждый следующий элемент не меньше предыдущего. Одинаковые значения допустимы.

Пусть `left` и `right` — индексы границ участка, включая обе границы. Средний индекс вычисляется целочисленным делением:

$$
middle=\left\lfloor\frac{left+right}{2}\right\rfloor.
$$

Скобки в формуле обозначают округление вниз; в Python такое деление целых чисел записывается через `//`.

Если средний элемент равен ключу, поиск завершён. Если он меньше ключа, все элементы левее него также не превосходят среднего и не могут быть искомым значением. Новая левая граница — `middle + 1`. Если средний элемент больше ключа, новая правая граница — `middle - 1`.

Когда `left > right`, участок пуст и ключ отсутствует. Уже проверенный средний элемент исключается из следующего участка, поэтому при каждой неудачной проверке область строго сокращается.

```python
def binary_search(values, target):
    left = 0
    right = len(values) - 1
    comparisons = 0
    while left <= right:
        middle = (left + right) // 2
        comparisons += 1
        if values[middle] == target:
            return middle, comparisons
        comparisons += 1
        if values[middle] < target:
            left = middle + 1
        else:
            right = middle - 1
    return -1, comparisons
```

При повторяющихся значениях эта реализация возвращает индекс **одного из вхождений**. Первый или последний индекс не гарантируется. Возвращаемый индекс относится к списку, переданному в функцию: после сортировки позиции элементов могут измениться.

Для непустого списка число итераций в худшем случае не превосходит $\lfloor\log_2n\rfloor+1$. На итерации выполняется одно сравнение при совпадении либо два при несовпадении. Поэтому число сравнений также имеет оценку $O(\log n)$. Лучший случай — совпадение с первым проверенным средним элементом: одно сравнение, $O(1)$. Для пустого списка сразу возвращается `(-1, 0)`.

Проверка отсортированности не включена в эту функцию. Условие на вход необходимо обеспечить до поиска. Случайный успешный запуск на неупорядоченном списке не подтверждает корректность такого применения.

### 1.6. Сортировка выбором

**Сортировка** располагает элементы в заданном порядке. При сортировке по неубыванию каждый следующий элемент не меньше предыдущего; одинаковые значения сохраняются в списке.

**Сортировка выбором** на позиции $i$ находит наименьший элемент оставшейся части и меняет его местами с элементом на этой позиции. После первого прохода на месте стоит наименьшее значение, после второго — два наименьших и так далее.

```python
def selection_sort(values):
    result = values.copy()
    comparisons = 0
    for i in range(len(result) - 1):
        smallest = i
        for j in range(i + 1, len(result)):
            comparisons += 1
            if result[j] < result[smallest]:
                smallest = j
        if smallest != i:
            result[i], result[smallest] = result[smallest], result[i]
    return result, comparisons
```

Функция возвращает новый список и число сравнений. Исходный список сохраняется, поскольку сортировка выполняется в его копии.

На первом проходе выполняется $n-1$ сравнений, на втором — $n-2$, на последнем — одно. Для $n\ge2$:

$$
C(n)=(n-1)+(n-2)+\ldots+1=\frac{n(n-1)}{2}=O(n^2).
$$

Эта сумма получается из $n-1$ слагаемых, среднее которых равно $n/2$. Порядок исходных элементов не меняет число сравнений: даже в уже отсортированном списке каждый проход просматривает оставшуюся часть. Для пустого списка и списка из одного элемента сравнений нет.

### 1.7. Сортировка вставками

**Сортировка вставками** поддерживает отсортированную начальную часть списка. Следующий элемент вставляется в подходящее место: большие значения сдвигаются вправо, а освободившаяся позиция заполняется текущим элементом.

```python
def insertion_sort(values):
    result = values.copy()
    comparisons = 0
    for i in range(1, len(result)):
        current = result[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if result[j] <= current:
                break
            result[j + 1] = result[j]
            j -= 1
        result[j + 1] = current
    return result, comparisons
```

Для уже упорядоченного непустого списка на каждом проходе достаточно сравнения с предыдущим элементом. Поэтому:

$$
C_{\text{лучший}}(n)=n-1=O(n).
$$

Для различных значений, расположенных в убывающем порядке, вставка очередного элемента требует просмотра всей начальной части:

$$
C_{\text{худший}}(n)=1+2+\ldots+(n-1)=\frac{n(n-1)}{2}=O(n^2).
$$

Одинаковая оценка худшего случая у двух сортировок ещё не означает одинаковое число операций на каждом входе. Для уже отсортированных данных вставки требуют меньше сравнений, чем выбор.

### 1.8. Стоимость подготовки данных

Бинарный поиск сокращает затраты на отдельный запрос, но сначала данные должны быть упорядочены. Если исходный список не отсортирован, сравнивают два полных способа решения:

$$
C_{\text{линейный}}=\sum_{j=1}^{q}C_{\text{линейного поиска},j},
$$

$$
C_{\text{с подготовкой}}=C_{\text{сортировки}}+
\sum_{j=1}^{q}C_{\text{бинарного поиска},j}.
$$

Здесь $q$ — количество запросов к одному набору данных. Если набор не меняется, сортировка выполняется один раз, а её стоимость распределяется между запросами. Для одного запроса подготовка может оказаться дороже последовательного поиска.

При сортировке выбором и $q$ запросах худшие оценки составляют $O(qn)$ для линейного поиска и $O(n^2+q\log n)$ для сортировки с последующим бинарным поиском. Утверждать, что один способ всегда лучше, без сведений о $n$, $q$ и состоянии входа нельзя. Другой алгоритм сортировки изменит стоимость подготовки.

Время выполнения можно измерять дополнительно, но отдельный замер зависит от среды и не доказывает асимптотическую оценку. Сначала устанавливают устройство алгоритма и число операций, затем используют эксперимент как проверку наблюдений.

## 2. Примеры

### Пример 1 Линейный поиск в неупорядоченном списке

**Задача.** В списке `[14, 3, 21, 8, 5, 13, 8, 2]` найти число 8, затем проверить наличие числа 9. Определить число сравнений.

**Решение.** Поиск числа 8 начинается с индекса 0:

| Индекс | Элемент | Результат сравнения с 8 | Накопленное число сравнений |
|---|---|---|---|
| 0 | 14 | Не равно | 1 |
| 1 | 3 | Не равно | 2 |
| 2 | 21 | Не равно | 3 |
| 3 | 8 | Равно: поиск завершён | 4 |

Получен индекс 3. Второе вхождение 8 на позиции 6 не проверяется, поскольку функция возвращает первое совпадение. Числа 9 в списке нет, поэтому последовательно проверяются все восемь элементов.

```python
values = [14, 3, 21, 8, 5, 13, 8, 2]
for target in (8, 9):
    index, comparisons = linear_search(values, target)
    print(target, "индекс:", index, "сравнения:", comparisons)
```

**Ответ.** Для 8: индекс 3, четыре сравнения. Для 9: индекс `-1`, восемь сравнений.

### Пример 2 Бинарный поиск и изменение границ

**Задача.** В упорядоченном списке `[2, 3, 5, 8, 8, 13, 14, 21]` найти 14. Показать изменение границ и подсчитать сравнения значений.

**Решение.** Исходные границы — 0 и 7.

| Итерация | `left` | `right` | Вычисление `middle` | Средний элемент | Действие | Сравнений на итерации |
|---|---|---|---|---|---|---|
| 1 | 0 | 7 | $(0+7)//2=3$ | 8 | $8\ne14$, $8<14$; `left = 4` | 2 |
| 2 | 4 | 7 | $(4+7)//2=5$ | 13 | $13\ne14$, $13<14$; `left = 6` | 2 |
| 3 | 6 | 7 | $(6+7)//2=6$ | 14 | $14=14$; поиск завершён | 1 |

Итого:

$$
C=2+2+1=5.
$$

Выполнено три итерации, но пять сравнений: на двух неудачных итерациях проверялись и равенство, и порядок. Индекс результата — 6 в упорядоченном списке. В исходном списке примера 1 число 14 находилось на позиции 0.

Проверим также отсутствие числа 9:

```python
ordered = [2, 3, 5, 8, 8, 13, 14, 21]
for target in (14, 9):
    index, comparisons = binary_search(ordered, target)
    print(target, "индекс:", index, "сравнения:", comparisons)
```

Для 9 последовательно проверяются значения 8, 13 и 8 на индексах 3, 5 и 4. После последней проверки `left = 5`, `right = 4`: участок пуст. Получается $C=2+2+2=6$ сравнений и индекс `-1`.

**Ответ.** Для 14: индекс 6, пять сравнений. Для 9: ключ отсутствует, шесть сравнений.

### Пример 3 Сортировки выбором и вставками

**Задача.** Упорядочить `[7, 2, 5, 2]` двумя способами и сравнить число сравнений. Проверить, как оно изменится для уже упорядоченного списка.

**Решение.** При сортировке выбором:

| Проход | Действие | Список после прохода | Сравнений |
|---|---|---|---|
| 1 | Наименьшее из четырёх значений — 2; обмен позиций 0 и 1 | `[2, 7, 5, 2]` | 3 |
| 2 | Наименьшее в оставшейся части — 2; обмен позиций 1 и 3 | `[2, 2, 5, 7]` | 2 |
| 3 | Наименьшее в оставшейся части — 5; обмен не нужен | `[2, 2, 5, 7]` | 1 |

$$
C_{\text{выбор}}=3+2+1=6.
$$

При вставках вначале отсортированная часть содержит только 7:

| Вставляемый элемент | Действие | Список после вставки | Сравнений |
|---|---|---|---|
| 2 | $7>2$; сдвинуть 7 и поставить 2 в начало | `[2, 7, 5, 2]` | 1 |
| 5 | $7>5$, затем $2\le5$; поставить 5 между 2 и 7 | `[2, 5, 7, 2]` | 2 |
| 2 | $7>2$, $5>2$, затем $2\le2$; вставить после первой 2 | `[2, 2, 5, 7]` | 3 |

$$
C_{\text{вставки}}=1+2+3=6.
$$

В программе сортировки вставками проверяется условие `result[j] <= current`: ложный результат означает, что элемент больше текущего и его нужно сдвинуть. В таблице тот же шаг описан через отношение «больше».

Для уже отсортированного списка `[2, 2, 5, 7]` выбор по-прежнему требует $3+2+1=6$ сравнений. Вставки на каждом из трёх проходов сразу находят подходящее место:

$$
C_{\text{вставки}}=1+1+1=3.
$$

```python
for values in ([7, 2, 5, 2], [2, 2, 5, 7], [7, 5, 2, 2]):
    selected, selection_count = selection_sort(values)
    inserted, insertion_count = insertion_sort(values)
    print("Исходный список:", values)
    print("Выбор:", selected, "сравнения:", selection_count)
    print("Вставки:", inserted, "сравнения:", insertion_count)
```

| Исходный список | Выбор | Вставки |
|---|---|---|
| `[7, 2, 5, 2]` | 6 | 6 |
| `[2, 2, 5, 7]` | 6 | 3 |
| `[7, 5, 2, 2]` | 6 | 6 |

**Ответ.** Обе сортировки получают `[2, 2, 5, 7]`. На первом входе счётчики совпали; на уже упорядоченном входе вставки выполнили вдвое меньше сравнений. Это не изменяет квадратичную оценку их худшего случая.

### Пример 4 Когда подготовка данных окупается

**Задача.** Для списка примера 1 сравнить линейный поиск и сортировку выбором с последующим бинарным поиском. Рассмотреть один запрос с ключом $-1$ и шестнадцать запросов с ключами от $-16$ до $-1$.

**Решение.** Все эти ключи меньше наименьшего элемента и отсутствуют в списке. Каждый линейный поиск выполняет восемь сравнений. Сортировка восьми элементов выбором требует:

$$
C_{\text{подготовки}}=\frac{8\cdot(8-1)}{2}=\frac{56}{2}=28.
$$

При бинарном поиске любого из указанных ключей проверяются элементы на позициях 3, 1 и 0: значения 8, 3 и 2. Каждое не совпадает с ключом и требует проверки порядка, поэтому:

$$
C_{\text{одного бинарного поиска}}=2+2+2=6.
$$

Для одного запроса:

$$
C_{\text{линейный}}=8,\qquad C_{\text{с подготовкой}}=28+6=34.
$$

Для шестнадцати запросов:

$$
C_{\text{линейный}}=16\cdot8=128,
$$

$$
C_{\text{с подготовкой}}=28+16\cdot6=28+96=124.
$$

Сортировка учитывается один раз для каждого из двух рассматриваемых сценариев.

```python
values = [14, 3, 21, 8, 5, 13, 8, 2]
ordered, preparation = selection_sort(values)
print("q, линейный, сортировка, бинарные запросы, сумма")
for q in (1, 16):
    queries = list(range(-q, 0))
    linear_count = sum(linear_search(values, key)[1] for key in queries)
    binary_count = sum(binary_search(ordered, key)[1] for key in queries)
    print(q, linear_count, preparation, binary_count,
          preparation + binary_count)
```

**Ответ.** Для одного запроса линейный способ требует меньше сравнений: 8 против 34. Для заданных шестнадцати запросов подготовка окупилась: 124 против 128. Вывод относится к указанным данным, ключам и выбранной сортировке; для другого набора затраты нужно пересчитать.

## 3. Проверка роста затрат

Пусть $n$ принимает значения 16, 64, 256 и 1024. Для поиска возьмём ключ $n$, которого нет в списке чисел от 0 до $n-1$. Линейный поиск проверит все элементы. В бинарном поиске каждый проверенный элемент меньше ключа, поэтому на каждой итерации выполняются два сравнения.

Для сортировки выбором порядок входа не влияет на число сравнений. Возьмём убывающий список и сопоставим счётчик с формулой $n(n-1)/2$.

```python
print("n, линейный поиск, бинарный поиск, сортировка выбором")
for n in (16, 64, 256, 1024):
    ordered = list(range(n))
    decreasing = list(range(n - 1, -1, -1))
    _, linear_count = linear_search(decreasing, n)
    _, binary_count = binary_search(ordered, n)
    result, selection_count = selection_sort(decreasing)
    assert result == ordered
    assert selection_count == n * (n - 1) // 2
    print(n, linear_count, binary_count, selection_count)
```

| $n$ | Линейный поиск | Бинарный поиск | Сортировка выбором |
|---|---|---|---|
| 16 | 16 | 10 | 120 |
| 64 | 64 | 14 | 2016 |
| 256 | 256 | 18 | 32640 |
| 1024 | 1024 | 22 | 523776 |

При увеличении $n$ в четыре раза линейный счётчик увеличился в четыре раза. В бинарном поиске добавились две итерации, то есть четыре сравнения. У квадратичной суммы отношение приближается к 16 при росте $n$; оно не равно 16 в точности из-за слагаемого $-n$ в формуле.

Полученная таблица согласуется с устройством алгоритмов: полный просмотр, деление области пополам и последовательность вложенных просмотров оставшихся частей. Именно эти свойства обосновывают оценки роста.
