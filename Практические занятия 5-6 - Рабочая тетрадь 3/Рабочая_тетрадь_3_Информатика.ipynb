## 1. Целые числа без знака
### Теоретический материал

Для представления чисел в ЭВМ используют **битовые наборы** — последовательности нулей и единиц фиксированной длины. Позиция в битовом наборе называется **разрядом**.

Для целых чисел без знака обычно выделяют $k=8,16,32$ или $64$ разряда. Чтобы получить компьютерную запись числа, переведите его в двоичную систему счисления и дополните слева нулями до $k$ разрядов.

### Пример. Число 26 в восьмиразрядном наборе

**Задача:** найти представление беззнакового целого числа $26_{10}$ при $k=8$.

**Решение:** последовательно делим число на 2.

| Делимое | Частное | Остаток |
|---:|---:|---:|
| 26 | 13 | 0 |
| 13 | 6 | 1 |
| 6 | 3 | 0 |
| 3 | 1 | 1 |
| 1 | 0 | 1 |

Читаем остатки снизу вверх: $26_{10}=11010_2$.
Дополняем запись слева до восьми разрядов.

**Ответ:** `00011010`.

## 2. Прямой, обратный и дополнительный коды
### Теоретический материал

Для целых чисел со знаком используют **прямой код (ПК)**, **обратный код (ОК)** и **дополнительный код (ДК)**.

Старший (левый) разряд является знаковым: `0` для неотрицательного числа, `1` для отрицательного. Положительные числа во всех трёх кодах записываются одинаково: знаковый ноль и двоичная запись числа, дополненная нулями до нужной длины.

Для отрицательного числа:

- **ПК:** знаковая единица и двоичная запись модуля в оставшихся разрядах.
- **ОК:** инвертируются все разряды положительного представления модуля, включая знаковый. Равнозначный способ: инвертировать все разряды ПК отрицательного числа, кроме знакового.
- **ДК:** к ОК отрицательного числа прибавляется единица. Операция выполняется в фиксированной разрядной сетке; перенос за её пределы отбрасывается.

В ОК и ДК младшие разряды отрицательного числа **не являются обычной двоичной записью модуля**.

В записи вида `1,0111000` запятая отделяет знаковый разряд и не означает дробную часть. Всего здесь восемь бит.

Для $k$ разрядов диапазон ПК и ОК: $-(2^{k-1}-1)\ldots 2^{k-1}-1$; диапазон ДК: $-2^{k-1}\ldots 2^{k-1}-1$. В ПК и ОК есть две записи нуля.

### Пример 1. Коды числа 45, k = 8

Последовательное деление на 2:

| Делимое | Частное | Остаток |
|---:|---:|---:|
| 45 | 22 | 1 |
| 22 | 11 | 0 |
| 11 | 5 | 1 |
| 5 | 2 | 1 |
| 2 | 1 | 0 |
| 1 | 0 | 1 |

Получаем $45_{10}=101101_2$. Дополняем модуль до семи разрядов: `0101101`. Знаковый разряд — `0`.

**Ответ:** ПК = ОК = ДК = `0,0101101`.

### Пример 2. Коды числа −56, k = 8

$56_{10}=111000_2$. Дополняем модуль до семи разрядов: `0111000`.

1. ПК: ставим знаковую единицу → `1,0111000`.
2. ОК: инвертируем младшие семь разрядов ПК → `1,1000111`.
3. ДК: прибавляем единицу к ОК: $11000111_2+1_2=11001000_2$.

| Код | Представление |
|---|---|
| Прямой | `1,0111000` |
| Обратный | `1,1000111` |
| Дополнительный | `1,1001000` |

## 3. Сложение и вычитание
### Теоретический материал

Сложение и вычитание чисел без знака осуществляется по стандартным алгоритмам для позиционных систем счисления.

**В обратном коде** складывают все разряды, включая знаковый. Если появился перенос за пределы $k$ разрядов, его прибавляют к младшему разряду результата. Это **циклический перенос**.

**В дополнительном коде** также складывают все разряды, включая знаковый. Перенос за пределы $k$ разрядов отбрасывают.

Вычитание сводят к сложению: $x-y=x+(-y)$.

Перед операцией выбирают общую разрядность, достаточную для операндов и результата. Перенос за пределы сетки и переполнение диапазона чисел со знаком — разные явления.

### Пример 1. Сложение 7 и 16

$7_{10}=111_2$, $16_{10}=10000_2$. Для этого примера используем шесть разрядов со знаком:

$[A]_{\text{ПК}}=[A]_{\text{ОК}}=[A]_{\text{ДК}}=0,00111$;

$[B]_{\text{ПК}}=[B]_{\text{ОК}}=[B]_{\text{ДК}}=0,10000$.

В ОК и ДК получаем одинаковый результат:

$$000111_2+010000_2=010111_2=23_{10}.$$

**Ответ:** 23.

### Пример 2. Сложение +16 и −7 в ОК и ДК

Используем шесть разрядов со знаком.

| Число | ПК | ОК | ДК |
|---|---|---|---|
| +16 | `0,10000` | `0,10000` | `0,10000` |
| −7 | `1,00111` | `1,11000` | `1,11001` |

**В ОК:** $010000_2+111000_2=1\,001000_2$.
За пределы шестиразрядной сетки вышла единица. Прибавляем её к младшему разряду: $001000_2+1_2=001001_2$.

**В ДК:** $010000_2+111001_2=1\,001001_2$.
Отбрасываем перенос за пределы сетки и получаем `001001`.

**Ответ:** $9_{10}$.

### 5.2. Пример 1. Вычисление выражения

Для введённых чисел $x$ и $y$ найдите

$$f(x,y)=2y^x+\ln|x+y^3|.$$

### 5.3. Пример 2. Кусочно заданная функция

Для введённых чисел $x$ и $y$ найдите

$$f(x,y)=\begin{cases}
\sin(xe^y), & xy\leq -1,\\
\sqrt{|\cos(xy)|}, & -1<xy<5,\\
x^2+\tan y, & xy\geq 5.
\end{cases}$$

Запустите программу с разными данными, чтобы проверить каждую ветвь. Например: $(x,y)=(-1,1)$, $(0,1)$, $(5,1)$.

```python
import math

x = float(input('Введите x: '))
y = float(input('Введите y: '))
if x * y <= -1:
    f = math.sin(x * math.exp(y))
elif x * y >= 5:
    f = x * x + math.tan(y)
else:
    f = math.sqrt(math.fabs(math.cos(x * y)))
print('f =', f)
```

### 5.4. Пример 3. Табулирование функции

Вычислите значения функции

$$f(x)=\sin(x-e^2)+3^x$$

на отрезке $[x_n,x_k]$ с шагом $h_x$.

Например, введите `0`, `1`, `0.2`.

Индекс шага позволяет вычислять $x=x_n+i h_x$ без последовательного накопления ошибки сложения дробных чисел.

```python
import math

xn = float(input('Введите xn: '))
xk = float(input('Введите xk: '))
hx = float(input('Введите hx: '))
if not all(math.isfinite(v) for v in (xn, xk, hx)):
    print('Введите конечные числа.')
elif hx <= 0 or xn > xk:
    print('Нужно hx > 0 и xn <= xk.')
else:
    i = 0
    x = xn
    while x <= xk or math.isclose(x, xk, rel_tol=1e-12, abs_tol=1e-12):
        f = math.sin(x - math.exp(2)) + math.pow(3, x)
        print(f'x = {x:.3f}, f = {f:.6f}')
        i += 1
        x = xn + i * hx
```

### 5.5. Пример 4*. Два вложенных цикла

Вычислите значения функции

$$f(x,y)=\begin{cases}
\sqrt[5]{y+x}, & x+y\leq 2,\\
|\sin x|^y, & x+y>2.
\end{cases}$$

$x$ изменяется от $0$ до $1$ с шагом $h_x=0{,}2$; $y$ — от $1$ до $2$ с шагом $h_y=0{,}5$.

Для каждого значения $x$ внутренний цикл проходит все значения $y$. В результате получается 18 строк.

```python
import math

ax, bx, hx = 0.0, 1.0, 0.2
ay, by, hy = 1.0, 2.0, 0.5

i = 0
x = ax
while x <= bx or math.isclose(x, bx, abs_tol=1e-12):
    j = 0
    y = ay
    while y <= by or math.isclose(y, by, abs_tol=1e-12):
        if x + y <= 2:
            f = math.pow(x + y, 1.0 / 5.0)
        else:
            f = math.pow(math.fabs(math.sin(x)), y)
        print(f'x = {x:.3f}, y = {y:.3f}, f = {f:.6f}')
        j += 1
        y = ay + j * hy
    i += 1
    x = ax + i * hx
```
