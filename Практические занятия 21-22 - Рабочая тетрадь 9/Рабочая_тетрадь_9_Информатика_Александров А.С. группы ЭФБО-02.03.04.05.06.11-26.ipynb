## 1. Теоретический материал

### 1.1. Итерационный алгоритм и последовательность значений

Алгоритм задаёт последовательность действий над исходными данными. В итерационном алгоритме одна и та же операция или группа операций выполняется многократно. Одно её выполнение называется **итерацией**.

Состояние алгоритма определяется текущими значениями его переменных. В числовом итерационном процессе правило перехода $T$ вычисляет новое значение $x$ по предыдущему. При начальном значении $x_0$ последовательно определяются:

$$
x_1=T(x_0),\qquad x_2=T(x_1),\qquad x_3=T(x_2).
$$

Общая запись имеет вид:

$$
x_{k+1}=T(x_k),\qquad k=0,1,2,\ldots
$$

Здесь $k$ — номер шага, $x_k$ — значение после $k$ шагов. Запись $x_0$ означает, что ещё не выполнено ни одного преобразования. **Траектория** — последовательность $x_0,x_1,x_2,\ldots$, полученная из заданного начального значения.

Например, для $T(x)=x+2$ и $x_0=-3$:

$$
x_1=T(-3)=-3+2=-1,
$$

$$
x_2=T(-1)=-1+2=1,
$$

$$
x_3=T(1)=1+2=3.
$$

Траектория: $-3\to-1\to1\to3\to5\to\ldots$.

Правило может состоять из нескольких ветвей. На каждом шаге условие проверяется **для текущего значения**. Выбирается одна подходящая ветвь, и выполняется одно преобразование. После изменения числа следующую ветвь проверяют уже на следующем шаге.

В Python такому правилу соответствует функция с `if` и `return`. Если последовательно записать два независимых `if` и изменять одну переменную, за один проход могут выполниться два преобразования. Это будет другой алгоритм.

### 1.2. Таблица трассировки

**Трассировка** — последовательная запись значений, возникающих при выполнении алгоритма. Для правила $T(x)=x+2$ таблица начинается так:

| Номер шага $k$ | Значение $x_k$ | Вычисление следующего значения |
|---|---|---|
| 0 | $-3$ | $-3+2=-1$ |
| 1 | $-1$ | $-1+2=1$ |
| 2 | $1$ | $1+2=3$ |
| 3 | $3$ | $3+2=5$ |

Таблица помогает проверить расчёты и заметить повтор. Однако несколько строк таблицы описывают только выбранный старт. Вывод для всех начальных значений требует отдельного обоснования.

### 1.3. Фиксированные точки

**Фиксированная, или неподвижная, точка** — допустимое значение $x^*$, для которого

$$
T(x^*)=x^*.
$$

Если траектория попала в фиксированную точку, все последующие значения равны ей. Например, для $T(x)=2x-4$:

$$
x=2x-4\quad\Longrightarrow\quad x-2x=-4\quad\Longrightarrow\quad -x=-4\quad\Longrightarrow\quad x=4.
$$

Проверка: $T(4)=2\cdot4-4=8-4=4$. Значит, 4 — фиксированная точка. Этот расчёт ещё не означает, что другие начальные значения обязательно придут к 4.

Для кусочного правила уравнение $T(x)=x$ решают отдельно в каждой ветви. Каждый найденный корень проверяют по условию соответствующей ветви. Если алгоритм задан над целыми числами, корень также должен быть целым.

Например, в ветви $T(x)=2x-4$ при $x<0$ уравнение снова даёт $x=4$. Но $4<0$ неверно, поэтому эта ветвь фиксированной точки не даёт. Если в некоторой ветви записано $T(x)=x$, фиксированными будут все допустимые значения этой ветви.

### 1.4. Цикл и его период

**Цикл** — повторяющийся участок траектории. **Период цикла** — число шагов, необходимое для одного полного обхода: наименьшее положительное число переходов, после которого происходит возврат к тому же значению и повторяется дальнейшая последовательность.

Пусть значение $x_i$ впервые повторилось на шаге $j$, то есть $x_j=x_i$ и $j>i$. Если правило зависит только от текущего числа, следующие значения также совпадут:

$$
x_{j+1}=T(x_j)=T(x_i)=x_{i+1}.
$$

По той же причине повторятся и все дальнейшие значения. Если это первый обнаруженный повтор в траектории, период равен разности номеров шагов:

$$
L=j-i.
$$

При подсчёте периода учитываются все переходы, включая последний переход к начальному значению. Повторное значение в конце записи не считается новым элементом цикла.

Например, для правила $T(x)=1-x$ и начального значения $x_0=0$:

$$
x_1=T(0)=1-0=1,
$$

$$
x_2=T(1)=1-1=0.
$$

Получается цикл $0\to1\to0$. Для возвращения к нулю нужны два перехода: из 0 в 1 и из 1 в 0. Через один шаг значение ещё отличается от начального, поэтому период равен $2-0=2$.

**Фиксированная точка — цикл периода 1.** Из равенства $T(x^*)=x^*$ следует, что уже один шаг оставляет значение прежним: $x^*\to x^*$. Один — наименьшее положительное число шагов, поэтому период равен 1. Например, если $T(0)=0$, цикл записывается как $0\to0$ и содержит один переход.

**Начальный участок** — значения траектории до первого попадания в цикл. Если начальное значение уже принадлежит циклу, этот участок пуст. Его длина не входит в период цикла.

### 1.5. Поведение при произвольном начальном значении

Полная характеристика поведения алгоритма должна охватывать **все допустимые начальные значения**. Траектории могут приводить к разным фиксированным точкам, разным циклам или неограниченному росту либо убыванию.

Порядок исследования:

1. Указывают область значений и находят фиксированные точки.
2. Строят несколько трасс, в том числе с отрицательными значениями и значениями на границах ветвей.
3. Выделяют области, в которых значение возрастает или убывает. Выясняют, может ли оно оставаться там бесконечно.
4. Если траектория попадает в конечное множество, проверяют переходы из каждого его элемента и находят все циклы.
5. Обосновывают, какие начальные значения приводят к каждому из найденных исходов.

Множество замкнуто относительно правила $T$, если из любого его элемента следующий шаг снова ведёт в это множество. В конечном замкнутом множестве траектория обязательно повторит значение: различных состояний конечное число. При этом циклов может быть несколько. Чтобы утверждать, что все старты приходят в один цикл, нужно проверить переходы и показать, что других циклов нет.

Для доказательства попадания в конечную область удобно найти целую неотрицательную величину, которая строго уменьшается, пока траектория находится вне этой области. Например, расстояние до нуля не может уменьшаться на 1 бесконечно, оставаясь положительным.

### 1.6. Исследование последовательности и завершение программы

Правило $x_{k+1}=T(x_k)$ само по себе не задаёт момента остановки. Даже после попадания в фиксированную точку или цикл последовательность можно продолжать.

Программа исследования может завершиться, когда обнаружит повтор. Это означает, что повторяющийся участок найден. Если программа остановилась после заданного числа шагов без повтора, известно только, что **за эти шаги повтор не обнаружен**. Причиной может быть как неограниченное движение, так и длинный начальный участок перед циклом.

Если переход зависит только от текущего числа, для обнаружения цикла достаточно найти повтор этого числа. Если переход зависит также от номера шага или других переменных, необходимо проверять повтор полного состояния алгоритма.

## 2. Примеры

### Пример 1 Попадание в фиксированную точку

**Задача.** Для всех целых начальных значений исследовать правило

$$
T_1(x)=\begin{cases}x-1,&x>0,\\0,&x=0,\\x+1,&x<0\end{cases}.
$$

Найти фиксированные точки, построить трассы из 3 и $-3$ и охарактеризовать поведение при произвольном $x_0$.

**Решение.** Для поиска фиксированных точек необходимо решить уравнение $T_1(x)=x$ в каждой ветви.

При $x>0$:

$$
x-1=x\quad\Longrightarrow\quad x-x=1\quad\Longrightarrow\quad0=1.
$$

Решений нет. При $x<0$:

$$
x+1=x\quad\Longrightarrow\quad x-x=-1\quad\Longrightarrow\quad0=-1.
$$

Решений также нет. При $x=0$ выполняется $T_1(0)=0$. Единственная фиксированная точка — 0.

Трассы из $x_0=3$ и $x_0=-3$:

| $k$ | Из $x_0=3$ | Из $x_0=-3$ |
|---|---|---|
| 0 | 3 | $-3$ |
| 1 | $3-1=2$ | $-3+1=-2$ |
| 2 | $2-1=1$ | $-2+1=-1$ |
| 3 | $1-1=0$ | $-1+1=0$ |
| 4 | 0 | 0 |

Для положительного старта каждый шаг уменьшает число на 1. Оно остаётся целым и через $x_0$ шагов достигает нуля. Для отрицательного старта каждый шаг прибавляет 1, поэтому через $-x_0$ шагов также получается ноль. При $x_0=0$ начальное значение уже фиксированное.

При $x_k\ne0$ модуль числа уменьшается на 1. Поэтому:

$$
x_{|x_0|}=0,\qquad x_k=0\text{ при }k\ge|x_0|.
$$

После достижения нуля каждый следующий шаг снова даёт ноль: $T_1(0)=0$. Цикл записывается как $0\to0$ и содержит один переход, поэтому его период равен 1. В таблице ноль впервые получен при $k=3$ и повторяется при $k=4$: $L=4-3=1$. Начальный участок первой траектории — $3,2,1$, второй — $-3,-2,-1$.

**Ответ.** Фиксированная точка — 0. Любая траектория достигает её через $|x_0|$ шагов и далее остаётся в нуле. Период цикла $0\to0$ равен 1.

### Пример 2 Цикл для любого целого начального значения

**Задача.** Исследовать правило над целыми числами:

$$
T_2(x)=\begin{cases}x-2,&x\ge2,\\x+3,&x<2\end{cases}.
$$

Найти фиксированные точки и охарактеризовать поведение при произвольном целом $x_0$.

**Решение.** Для первой ветви:

$$
x-2=x\quad\Longrightarrow\quad x-x=2\quad\Longrightarrow\quad0=2.
$$

Для второй ветви:

$$
x+3=x\quad\Longrightarrow\quad x-x=-3\quad\Longrightarrow\quad0=-3.
$$

Оба равенства невозможны. Фиксированных точек нет.

Трасса из $x_0=0$:

| $k$ | $x_k$ | Выбранная ветвь | Следующее значение |
|---|---|---|---|
| 0 | 0 | $x<2$ | $0+3=3$ |
| 1 | 3 | $x\ge2$ | $3-2=1$ |
| 2 | 1 | $x<2$ | $1+3=4$ |
| 3 | 4 | $x\ge2$ | $4-2=2$ |
| 4 | 2 | $x\ge2$ | $2-2=0$ |
| 5 | 0 | Повтор значения при $k=0$ | Начинается следующий обход |

Ноль впервые повторился на шаге 5, поэтому цикл имеет вид $0\to3\to1\to4\to2\to0$. Между двумя появлениями нуля выполнено пять переходов: $L=5-0=5$. Значения на шагах 1–4 равны 3, 1, 4 и 2; ни одно не равно нулю. Следовательно, вернуться к началу за меньшее положительное число шагов нельзя, и период равен именно 5.

Целые начальные значения разделяются на три случая.

**Случай $x_0\ge2$.** Пока число не меньше 2, из него вычитается 2. Вычитание 2 сохраняет чётность числа. Чётное число за конечное число шагов достигает 0, нечётное — 1. Оба значения принадлежат найденному циклу.

**Случай $x_0<0$.** Пока число отрицательно, к нему прибавляется 3. Оно за конечное число шагов впервые становится неотрицательным. Последнее отрицательное значение перед этим переходом не меньше $-3$: иначе прибавление 3 оставило бы его отрицательным. Поэтому оно равно $-3$, $-2$ или $-1$, а первым неотрицательным значением будет 0, 1 или 2. Каждое из этих значений входит в цикл.

**Случаи $x_0=0$ и $x_0=1$.** Старт уже находится в цикле.

Эти случаи охватывают все целые начальные значения. Множество $\{0,1,2,3,4\}$ замкнуто, а таблица переходов показывает, что все пять элементов образуют один цикл.

Например, при $x_0=-3$:

$$
-3\to0\to3\to1\to4\to2\to0\to\ldots
$$

Начальный участок — $-3$, первое попадание в цикл происходит при $k=1$. При старте из 3 тот же цикл записывается как $3\to1\to4\to2\to0\to3$. Выбор начального элемента в записи не создаёт нового цикла.

**Ответ.** Фиксированных точек нет. При любом целом $x_0$ траектория за конечное число шагов входит в цикл $0\to3\to1\to4\to2\to0$ периода 5.

### Пример 3 Разные исходы для разных начальных значений

**Задача.** Для всех целых $x_0$ исследовать правило

$$
T_3(x)=\begin{cases}x+1,&x<0,\\0,&x=0,\\3-x,&1\le x\le2,\\x+1,&x\ge3\end{cases}.
$$

Найти фиксированные точки, циклы и области начальных значений с различным поведением.

**Решение.** В ветвях $x<0$ и $x\ge3$ уравнение $x+1=x$ приводит к $0=-1$, поэтому фиксированных точек там нет. При $x=0$ выполняется $T_3(0)=0$.

В ветви $1\le x\le2$:

$$
3-x=x\quad\Longrightarrow\quad3=2x\quad\Longrightarrow\quad x=\frac32.
$$

Число $3/2$ не целое, поэтому не относится к допустимым значениям. Единственная фиксированная точка — 0.

Поведение последовательности зависит от начального значения.

При $x_0<0$ число увеличивается на 1 и через $-x_0$ шагов достигает нуля. Затем каждый шаг снова даёт ноль, поскольку $T_3(0)=0$. Это цикл $0\to0$ периода 1. Например:

$$
-2\to-1\to0\to0\to\ldots
$$

При $x_0=0$ траектория сразу постоянна.

Для двух допустимых значений из третьей ветви:

$$
T_3(1)=3-1=2,\qquad T_3(2)=3-2=1.
$$

Значения 1 и 2 образуют цикл $1\to2\to1$. За один шаг значение меняется, а за два возвращается к исходному, поэтому период равен 2. Переходы из обоих элементов ведут друг в друга, следовательно, выйти из этого цикла нельзя.

При $x_0\ge3$ каждое следующее значение увеличивается на 1 и остаётся не меньше 3. Поэтому на каждом шаге применяется последняя ветвь:

$$
x_1=x_0+1,
$$

$$
x_2=x_1+1=(x_0+1)+1=x_0+2.
$$

За $k$ шагов к исходному значению прибавляется $k$, следовательно:

$$
x_k=x_0+k.
$$

При увеличении $k$ значение может превысить любую заданную границу. Значит, последовательность неограниченно возрастает. Каждый новый элемент больше предыдущего, поэтому повторов нет. Например, $3\to4\to5\to6\to\ldots$.

**Ответ.** Все целые начальные значения распределяются так:

| Начальное значение | Поведение | Период после попадания в цикл |
|---|---|---|
| $x_0\le0$ | Достигается фиксированная точка 0 | 1 |
| $x_0\in\{1,2\}$ | Цикл $1\to2\to1$ | 2 |
| $x_0\ge3$ | Неограниченное возрастание | Цикла нет |

Одна трасса здесь не описывает весь алгоритм: запуск из отрицательного числа не покажет цикл из 1 и 2, а запуск из 1 не покажет неограниченного роста.

## 3. Проверка на Python

### 3.1. Функции перехода

Для проверки достаточно стандартного Python 3. Выполняйте кодовые ячейки этого раздела сверху вниз; устанавливать дополнительные библиотеки не требуется.

Первая кодовая ячейка содержит правила из разобранных примеров. Каждая функция получает текущее целое число и возвращает значение после одного шага.

```python
def step_fixed(x):
    if x > 0:
        return x - 1
    if x < 0:
        return x + 1
    return 0


def step_cycle(x):
    if x >= 2:
        return x - 2
    return x + 3


def step_mixed(x):
    if x < 0:
        return x + 1
    if x == 0:
        return 0
    if x <= 2:
        return 3 - x
    return x + 1
```

Вызов `step_cycle(-3)` возвращает 0. Чтобы получить следующий элемент траектории, функцию нужно вызвать ещё раз, теперь с аргументом 0. Команда `return` завершает текущий вызов, поэтому один вызов выполняет ровно одну ветвь правила.

### 3.2. Обнаружение повтора

Список `path` хранит значения в порядке появления. Словарь `seen` сопоставляет каждому уже встречавшемуся значению номер его первого появления. Если текущее значение уже есть в словаре, начальный участок и цикл можно выделить по этому номеру.

В функции `analyze` параметр `step` — выбранная функция перехода, `x0` — начальное число, `max_steps` — наибольшее число разрешённых преобразований. Например, `analyze(step_cycle, -3, 20)` исследует правило второго примера из $-3$ и допускает до 20 шагов.

```python
def analyze(step, x0, max_steps=100):
    if max_steps < 0:
        raise ValueError("Число шагов должно быть неотрицательным")
    path = []
    seen = {}
    x = x0
    for k in range(max_steps + 1):
        if x in seen:
            start = seen[x]
            return {
                "status": "repeat",
                "path": path + [x],
                "prefix": path[:start],
                "cycle": path[start:],
                "start": start,
                "period": k - start,
                "steps": k,
            }
        seen[x] = k
        path.append(x)
        if k < max_steps:
            x = step(x)
    return {
        "status": "limit",
        "path": path,
        "steps": max_steps,
    }
```

Функция возвращает словарь с результатом. При обнаружении повтора `path` включает замыкающее повторное значение; `cycle` содержит каждый элемент одного обхода ровно один раз.

| Поле | Содержание |
|---|---|
| `status` | `repeat` — найден повтор; `limit` — достигнут предел шагов |
| `path` | Просмотренные значения, начиная с $x_0$ |
| `steps` | Число выполненных преобразований |
| `prefix` | Начальный участок до первого попадания в цикл |
| `cycle` | Значения одного обхода цикла |
| `start` | Номер первого элемента цикла в траектории |
| `period` | Период цикла |

Последние четыре поля присутствуют только при `status == "repeat"`. Срез `path[:start]` берёт элементы до начала цикла, а `path[start:]` — от его начала до конца сохранённого списка.

При пределе $L$ рассматриваются значения $x_0,\ldots,x_L$, то есть $L+1$ состояний после не более $L$ преобразований. Если повтор возник ровно на последнем разрешённом шаге, функция его обнаружит. При `max_steps=0` проверяется только начальное значение.

### 3.3. Запуск и проверка результата

Выполните ячейку с выводом результатов и запусками четырёх примеров:

```python
def show(result):
    print("Трасса:", " -> ".join(map(str, result["path"])))
    print("Выполнено шагов:", result["steps"])
    if result["status"] == "repeat":
        print("Начальный участок:", result["prefix"])
        print("Цикл:", result["cycle"])
        print("Начало цикла k =", result["start"])
        print("Период:", result["period"])
    else:
        print("Достигнут предел шагов. Повтор не обнаружен.")


if __name__ == "__main__":
    examples = [
        ("Фиксированная точка", step_fixed, -3, 10),
        ("Периодический цикл", step_cycle, -3, 20),
        ("Разное поведение", step_mixed, 1, 10),
        ("Предел шагов", step_mixed, 3, 5),
    ]
    for title, step, x0, limit in examples:
        print(title)
        show(analyze(step, x0, limit))
        print()
```

| $T_3$, $x_0=3$, предел 5 | $3\to4\to5\to6\to7\to8$ | Повтор не найден | Не определён |

В последнем запуске программа сообщает только об отсутствии повтора за пять шагов. Неограниченное возрастание следует из формулы $x_k=x_0+k$ в примере 3.
